```python
# 值为6
from scapy.all import *

# 定义目标IPv6地址
target_ip = "2001:da8:a000:ab23::10"

# 构建IPv6报文
ipv6_packet = IPv6(dst=target_ip)

# 构建一个TCP报文
# 这里的Next Header为6，表示TCP协议
# 设置TCP的一些基本字段，例如目标端口80，源端口随机生成
tcp_packet = TCP(dport=80, sport=RandShort(), flags="S", seq=1000)

# 将IPv6包和TCP包结合起来
packet = ipv6_packet / tcp_packet

# 发送数据包
send(packet)

print(f"IPv6 packet with Next Header set to 6 (TCP) sent to {target_ip}")
```

```python
# 50
from scapy.all import *
import struct

# 定义目标IP地址
dst_ip = "2001:da8:a000:ab23::10"

# 创建一个IPv6报文并设置Next Header为50 (ESP协议)
ipv6_pkt = IPv6(dst=dst_ip, nh=50)

# 假设我们添加一个简单的ESP负载（这是一个虚拟负载，仅作为示例）
# 实际上，ESP负载应包括加密数据和可能的认证标记
esp_payload = Raw(load="ESP Payload Placeholder")

# 合并IPv6报文和ESP负载
pkt = ipv6_pkt / esp_payload

# 发送报文
send(pkt)

# 验证报文字段是否符合要求
def verify_packet(pkt):
    # 验证IPv6报文的Next Header值是否为50
    if pkt[IPv6].nh == 50:
        print("Next Header is correctly set to 50 (ESP).")
    else:
        print(f"Error: Next Header is {pkt[IPv6].nh}, expected 50.")
    
    # 验证目的地址是否正确
    if pkt[IPv6].dst == dst_ip:
        print(f"Destination address is correctly set to {dst_ip}.")
    else:
        print(f"Error: Destination address is {pkt[IPv6].dst}, expected {dst_ip}.")

# 验证构造的报文
verify_packet(pkt)
```

```python
# 51
from scapy.all import *
import struct

# 目标IP地址
dst_ip = "2001:da8:a000:ab23::10"

# 构造IPv6报文，设置Next Header为51（AH协议）
ipv6_pkt = IPv6(dst=dst_ip, nh=51)

# AH协议通常会带有特定的认证字段（如Sequence Number，SPI等），但Scapy不提供完整的AH构造支持。
# 为了示例，我们只加入一个占位的原始负载部分
ah_payload = Raw(load="AH Payload Placeholder")

# 合并IPv6报文和AH负载
pkt = ipv6_pkt / ah_payload

# 发送报文
send(pkt)

# 验证构造的报文字段
def verify_packet(pkt):
    # 验证IPv6报文的Next Header值是否为51（AH协议）
    if pkt[IPv6].nh == 51:
        print("Next Header is correctly set to 51 (AH).")
    else:
        print(f"Error: Next Header is {pkt[IPv6].nh}, expected 51.")
    
    # 验证目的地址是否正确
    if pkt[IPv6].dst == dst_ip:
        print(f"Destination address is correctly set to {dst_ip}.")
    else:
        print(f"Error: Destination address is {pkt[IPv6].dst}, expected {dst_ip}.")
    
    # 验证AH负载部分是否存在
    if len(pkt[Raw].load) > 0:
        print(f"AH payload is present: {pkt[Raw].load}")
    else:
        print("Error: AH payload is missing.")

# 验证构造的报文
verify_packet(pkt)
```

```python
# 60
from scapy.all import *

# 目标IPv6地址
dst_ip = "2001:da8:a000:ab23::10"

# 构造Destination Options Header（Next Header = 60）
# Destination Options Header一般会包含一个选项字段，这里使用一个简单的示例
# 需要指定头部长度和选项内容，我们可以简单地用Raw载荷表示这些内容

# 构造IPv6报文，设置Next Header为60（Destination Options Header）
ipv6_pkt = IPv6(dst=dst_ip, nh=60)

# 构造Destination Options Header（仅做示例）
# Destination Options Header通常包含Option Type, Option Length等信息
# 这里我们将它模拟为一个简单的Raw负载，实际应用中可能更复杂
dest_options_payload = Raw(load="Destination Options Header Payload")

# 合并IPv6报文和Destination Options Header负载
pkt = ipv6_pkt / dest_options_payload

# 发送报文
send(pkt)

# 验证构造的报文字段
def verify_packet(pkt):
    # 验证IPv6报文的Next Header值是否为60（Destination Options Header）
    if pkt[IPv6].nh == 60:
        print("Next Header is correctly set to 60 (Destination Options Header).")
    else:
        print(f"Error: Next Header is {pkt[IPv6].nh}, expected 60.")
    
    # 验证目的地址是否正确
    if pkt[IPv6].dst == dst_ip:
        print(f"Destination address is correctly set to {dst_ip}.")
    else:
        print(f"Error: Destination address is {pkt[IPv6].dst}, expected {dst_ip}.")
    
    # 验证Destination Options Header负载是否存在
    if len(pkt[Raw].load) > 0:
        print(f"Destination Options Header payload is present: {pkt[Raw].load}")
    else:
        print("Error: Destination Options Header payload is missing.")

# 验证构造的报文
verify_packet(pkt)
```
