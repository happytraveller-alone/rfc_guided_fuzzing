```python
from scapy.all import *

# 目标DNS服务器IP地址
dns_server = "192.168.115.155"

# 构造DNS查询报文
# 创建IP层和UDP层
ip = IP(dst=dns_server)
 
# DNS查询层，设置Opcode为0（标准查询）
dns_query = DNS(rd=1, qd=DNSQR(qname="www.example.com", qtype="A"))

# 合并IP层、UDP层和DNS查询层
pkt = ip / UDP(sport=RandShort(), dport=53) / dns_query

# 发送DNS查询报文
sr(pkt)
```

```python
from scapy.all import *

# Function to send DNS packet with opcode 2
def send_dns_packet():
    # Define the target IP address
    target_ip = "192.168.115.155"

    # Craft the DNS query packet with opcode 2 (Status Query)
    dns_packet = IP(dst=target_ip) / UDP(dport=53, sport=RandShort()) / DNS(
        id=RandShort(),  # Random DNS ID
        qr=0,             # Query (not response)
        opcode=2,         # Status (opcode 2)
        qdcount=0,        # No queries
        ancount=0,        # No answers
        nscount=0,        # No authority
        arcount=0         # No additional records
    )

    # Send the crafted packet
    send(dns_packet)
    print(f"DNS packet with opcode 2 sent to {target_ip}")

# Send the DNS packet
send_dns_packet()
```
